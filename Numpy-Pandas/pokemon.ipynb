# Pandas

Vamos a revisar los datos de distintos Pokemon con el dataset [Pokemon with stats](https://www.kaggle.com/datasets/abcsds/pokemon?resource=download)

### Importamos librerias y cargamos el archivo

```python
# Nota para tutor: Usar python 3.11.13 xd
import pandas as pd

pokemon = pd.read_csv("Pokemon.csv")
print(pokemon.info())
print(type(pokemon))
print(pokemon.head())
print(pokemon.columns)
print(pokemon.shape)
```

### 1. Revisar datos de interes

```python
tipos = pd.concat([pokemon["Type 1"], pokemon["Type 2"].dropna()])
print("Tipos distintos que hay", tipos.unique(), "Con cantidad", tipos.nunique())
```

```python
generaciones = pokemon["Generation"]
print("Numero de generaciones:", generaciones.unique(), "con cantidad", generaciones.nunique())
```

```python
print(pokemon.describe())
```

### 2. Revisar todos los pokemon de tipo normal

```python
normales = pokemon[(pokemon["Type 1"] == "Normal") | (pokemon["Type 2"] == "Normal")]
print(normales)
print(normales.shape)
```

### 3. Revisar todos los pokemon que solo sean tipo planta

```python
solo_planta = pokemon[(pokemon["Type 1"] == "Grass") & (pokemon["Type 2"].isna())]
print(solo_planta)
print(solo_planta.shape)
```

### 4- Obtener todos los legendarios y ordenarlos por HP

```python
legendarios = pokemon[pokemon["Legendary"] == True].sort_values(by="HP", ascending=False)
print(legendarios)
print(legendarios.shape)
```

### 5. Sacar la media de ataque fisico por generación

```python
media = pokemon.groupby("Generation")["Attack"].mean().sort_values(ascending=False)
print(media)
```

### 6. Obtener los pokemon de tipo fuego con velocidad mayor a 100, que no sean legendarios y ordenarlos por su defensa de mayor a menor. Mostrar solamente el nombre, su defensa y la generación. 

```python
pokemon_filtrado = pokemon[
    ((pokemon["Type 1"] == "Fire") | (pokemon["Type 2"] == "Fire")) &
    (pokemon["Speed"] > 100) &
    (pokemon["Legendary"] == False)
][["Name", "Defense", "Generation"]].sort_values(by="Defense", ascending=False)

print(pokemon_filtrado)
print(pokemon_filtrado.shape)
```
